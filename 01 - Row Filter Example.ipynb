```python
%sql
USE CATALOG dbacademy;
USE SCHEMA default;
```

## Create a UDF (User Function)

```python
%sql
CREATE OR REPLACE FUNCTION dbacademy.default.us_only(region STRING)
RETURN CASE 
    WHEN region = 'US' AND is_account_group_member('admins') THEN TRUE
    ELSE FALSE
END;
```

## Create Table and Data

```python
%sql
CREATE TABLE IF NOT EXISTS sales (
  id INT,
  region STRING,
  amount DECIMAL(12,2)
);
```

## ADD DATA

```python
%sql
INSERT INTO sales
VALUES (1, 'US', 10.00), (2, 'EU',20.00),(3,'US', 30.00);
```

## Attach the filter

```python
%sql
ALTER TABLE sales SET ROW FILTER dbacademy.default.us_only ON (region);
```
