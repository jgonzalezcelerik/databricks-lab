```python
my_catalog = dbutils.widgets.get('catalog')
my_schema = dbutils.widgets.get('schema')

# Set the path of your volume
my_volume_path = f"/Volumes/{my_catalog}/{my_schema}/trigger_storage_location/"

# Load the customer.csv file from the volume
customer_df = (spark
               .read
               .format("csv")
               .option("header", "true")
               .load(my_volume_path)
               )
customer_df.write.mode("overwrite").saveAsTable(f'{my_catalog}.{my_schema}.customer_bronze')
display(customer_df)
```
