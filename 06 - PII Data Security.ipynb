```python
salt = "BEANS"

def salted_hash (id):
    return F.sha2(F.concat(id, F.lit(salt)), 256)
```

```python
@dtl.table
def user_lookup_hashed():
    return (
        dtl
        .read_stream("people_10m")
        .select(
            salted_hash(F.col('id')).alias('alt_id'),
            "id",
            "firstName",
            "lastName"
        )
    )
```
